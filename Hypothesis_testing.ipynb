```python
from scipy.stats import binom
```

```python
n = 10
p  = 0.5
ob_heads = 2
alpha = 0.05

#pvalue < alpha , reject H0
#pvalue = p(head <= 2) 
pvalue = binom.cdf(ob_heads ,n, p)
         
```

```python
pvalue 
```

```python
if pvalue < alpha :
    print('Reject H0')
else:
    print('Do not reject H0')
```
